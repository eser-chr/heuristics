```python
import pandas as pd

from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
base = Path("./results/hypertuning/alpha")
```

```python
df = pd.DataFrame()
for subdir in base.iterdir():
    tmp = pd.read_csv(subdir)
    N = int(subdir.stem)
    plt.plot(tmp["alpha"], tmp["objective"])
    tmp["N"] = N
    df = pd.concat([df, tmp])
```

```python

df.sort_values(by = "N")
df = df[["alpha", "objective", "N"]]
```

```python
df
```

```python
df.groupby(by = ["N", "alpha"]).mean()
```

```python
fig, ax = plt.subplots(1,1)
sns.lineplot(df, x = "alpha", y = "objective", hue = "N", ax= ax)
ax.set(yscale = "log")
```
